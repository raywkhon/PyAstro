There are two methods for containing solution results:
[dictionary](#solve-and-use-results-in-a-dictionary) or [set](#solve-results-in-a-set). A dictionary is easier to interrogate
programmatically, so if you need to extract solutions using code, we recommend the [dictionary](#solve-and-use-results-in-a-dictionary) approach.


Solve and use results in a Dictionary - You can solve a system of equations for some variables leaving another symbol as a constant or parameter. You can specify the variables to solve for as multiple separate arguments, or as a list (or tuple):

```python
from sympy import solve
from sympy.abc import x, y, z
equations = [x**2 + y - 2*z, y + 4*z]
equations
```

```python
solutions = solve(equations, x, y, dict=True) # get a list of dictionaries with `dict=True`
solutions
```

Use a solution given as a dictionary - extract solutions by indexing (specifying in brackets) the solution number, and then the symbol. For example `solutions[0][x]` gives the result for `x` in the first solution:

```python
print(solutions[0][x])
solutions[1][x]
```

```python
print(solutions[0][y])
solutions[1][y]
```

Solve results in a set - to get a list of symbols and set of solutions, use `set=True` instead of `dict=True`:

```python
from sympy import solve
from sympy.abc import x, y, z
solve([x**2 + y - 2*z, y + 4*z], [x, y], set=True) # get a set of solutions, with `set=True`
```

Not all systems of equations can be solved - ystems of Equations With no Solution, e.g., the following two systems have no solution because they reduce to `1 == 0`, so SymPy returns an empty list:

```python
from sympy import solve
from sympy.abc import x, y
solve([x + y - 1, x + y], [x, y], set=True)
```

```python
from sympy import solve
from sympy.abc import x, y
solve([x + y - 1, x + y], [x, y], dict=True)
```

The following system reduces to $z = 2z$, so it has no general solution, but it could be satisfied if $z=0$. Note that {func}`~.solve` will not assume that $z=0$, even though that is the only value of $z$ that makes the system of equations consistent, because $z$ is a parameter rather than an unknown. That
is, {func}`~.solve` does not treat $z$ as an unknown because it is not in the list of symbols specified as unknowns (`[x, y]`) and all such symbols are treated like parameters with arbitrary value. Whether a symbol is treated as a variable or a parameter is determined only by whether it is specified as a symbol to solve for in {func}`~.solve`. There is no such distinction made when creating the symbol using {func}`~.symbols` (or importing from {mod}`~.abc`).

```python
Sol = [x + y - z, x + y - 2*z]
Sol
```

```python
from sympy import solve
from sympy.abc import x, y, z
solve([x + y - z, x + y - 2*z], [x, y], dict=True)
```

The following system is [overconstrained](https://en.wikipedia.org/wiki/Overdetermined_system), meaning there are more equations (three) than unknowns to be solved for (two, namely $x$ and $y$). It has no solution:

```python
Sol = [x + y - z, x - (z + 1), 2*x - y]
Sol
```

```python
from sympy import solve
from sympy.abc import x, y, z
solve([x + y - z, x - (z + 1), 2*x - y], [x, y], dict=True)
```

Note that some overconstrained systems do have solutions (for example, if an equation is a linear combination of the others), in which case SymPy can solve the overconstrained system.

Systems of equations with no closed-form solution - Some systems of equations cannot be solved algebraically (either at all or by SymPy), so you may have to [solve your system of equations numerically](solve-numerically.md) using {func}`~.nsolve` instead. For example those containing [transcendental equations](https://en.wikipedia.org/wiki/Transcendental_equation):

```python
from sympy import cos, solve
from sympy.abc import x, y, z
Sol = [x - y, cos(x) - y]
Sol
```

```python
from sympy import cos, solve
from sympy.abc import x, y, z
solve([x - y, cos(x) - y], [x, y], dict=True) #  equations cannot be solved algebraically, so there will be NotImplementedError 
```

\use {func}`~.nsolve` to [find a numerical solution](solve-numerically.md):

```python
Equ = [x - y, cos(x) - y]
Equ
```

```python
from sympy import cos, nsolve
from sympy.abc import x, y, z
nsolve([x - y, cos(x) - y], [x, y], [1,1])
```
