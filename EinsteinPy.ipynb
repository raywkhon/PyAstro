Solving the 2 coupled differential equations for the neutron stars, i.e. the "mass continuity equation and the "Tolman-Oppenheimer-Volkoff (TOV) equation" (in geometrized units):<br>
1. Specify an equation of state (EOS) $p(\rho)$ relating pressure p and energy density $\rho$.<br>
2. Integrate numerically from the center $(r = 0)$ to the surface $(p = 0)$, using appropriate boundary conditions.<br>
3. Equations to Solve<br>
$dm/dr = 4\pi*r^2*\rho$ (Mass continuity)<br>
$dp/dr = \frac{-(\rho + p)(m + 4\pi*r^3*p)}{r(r - 2m)}$ (TOV equation)<br>
4. Boundary Conditions<br>
At the center $r = 0, m(0) = 0$ (no mass enclosed at $r = 0, p(0) = p_c$ (central pressure, chosen based on the star's properties).<br>
At the surface $r = R: p(R) = 0$ (pressure drops to zero at the surface).<br>
5. Equation of State (EOS)<br>
A simple but physically relevant EOS is the "polytropic equation of state":<br>
$p = K*\rho^\Gamma$, where: $K$ is a constant, $\Gamma$ is the adiabatic index (e.g., $\Gamma = 2$ for neutron stars).<br>
For a more realistic model, tabulated EOS data (e.g., SLy, APR) can be used.
6. Numerical Integration (Pseudocode)<br>
We use the Runge-Kutta 4th-order (RK4) method to integrate the equations outward from $r = 0$:<br>
Example : for a neutron star with:<br>
$p_c = 10^{35} Pa, \rho_c = 5 * 10^{17} kg/m^3, \Gamma = 2, K = 1 * 10^5$<br>
You would obtain:<br>
Mass M approx = 1.4 solar masses and radius R approx = 10 km.<br>

```python
import numpy as np
from sympy import *

def TOV_solver(p_c, rho_c, Gamma, K, r_max, dr):
    # Initialize
    n = 0
    r = 0.01
    m = 4/3 * np.pi * r**3 * rho_c
    # p = K * rho_c ** Gamma
    p = p_c
    rho = rho_c
    dp_dr = - (rho + p) * (m + 4 * np.pi * r**3 * p) / (r * (r - 2 * m) )
    print(f"\nIteration = {n}")
    print(f"rho = {rho_c}")
    print(f"r = {r}")
    print("m = {:3e}".format(m))
    print("r - 2m ={:3e}".format(r - 2*m))
    print("p = {:3e}".format(p))
    print("dp/dr ={:3e}".format(dp_dr))
    
    # Arrays to store results
    r_list = [r]
    m_list = [m]
    p_list = [p]
    rho_list = [rho]
    
    # Integration loop

    while r < r_max and p > 0:
        # Compute derivatives
        n += 1
        r += dr
        p += dp_dr * dr
        rho = (p / K)**(1 / Gamma)
        dm_dr = 4 * np.pi * r**2 * rho
        m += dm_dr * dr
        dp_dr = - (rho + p) * (m + 4 * np.pi * r**3 * p) / (r * (r - 2 * m) )
        if n < 5 or n > 38800:
            print(f"\nIteration = {n}")
            print(f"rho = {rho}")
            print(f"r = {r}")
            print("m = {:3e}".format(m))
            print("r - 2m ={:3e}".format(r- 2*m))
            print("p = {:3e}".format(p))
            print("dp/dr ={:3e}".format(dp_dr))
       
        # Store results
        r_list.append(r)
        m_list.append(m)
        p_list.append(p)
        rho_list.append(rho)

    print("r_list, m_list, p_list, rho_list:")

    return 
    # return r_list, m_list, p_list, rho_list

TOV_solver(p_c = (10**35)*(8.2624*10**(-45)), rho_c = (5*10**17)*(7.42592*10**(-28)), Gamma = 2, K = 1*10**5, r_max=10**4, dr=0.01)
```

Implement the RK4 method

```python
import numpy as np
from sympy import *
import matplotlib.pyplot as plt

def rk4_integrate_TOV(p_c, rho_c, Gamma, K, r_max, dr):
    # Initialize
    n = 0
    r = 0.01
    m = 4/3 * np.pi * r**3 * rho_c
    # p = K * rho_c ** Gamma
    p = p_c
    rho = rho_c
    dp_dr = - (rho + p) * (m + 4 * np.pi * r**3 * p) / (r * (r - 2 * m) )
    print(f"\nIteration = {n}")
    print("rho ={:3e}".format(rho_c))
    dm_dr = 4 * np.pi * r**2 * rho
    print("dm_dr ={:3e}".format(dm_dr))
    print("dp_dr ={:3e}".format(dp_dr))
    print(f"r = {r}")
    print("m = {:3e}".format(m))
    print("p = {:3e}".format(p))
    
    # Arrays to store results
    r_list = [r]
    m_list = [m]
    p_list = [p]
    rho_list = [rho]
        
    def equations(r, m, p):
        rho = (p / K)**(1 / Gamma)
        dm_dr = 4 * np.pi * r**2 * rho
        #print(f"rho = {rho}")
        #print(f"dm_dr = {dm_dr}")
        dp_dr = - (rho + p) * (m + 4 * np.pi * r**3 * p) / (r * (r - 2 * m))
        #print(f"dp_dr = {dp_dr}")
        return dm_dr, dp_dr

    # RK4 integration loop
    while r < r_max and p > 0:
        n += 1
        #print(f"\nIteration = {n}")
    
        # Step 1: Compute k1
        #print(f"Step 1:")
        dm_dr_k1, dp_dr_k1 = equations(r, m, p)
        k1_m = dr * dm_dr_k1
        k1_p = dr * dp_dr_k1

        # Step 2: Compute k2
        #print(f"Step 2:")
        dm_dr_k2, dp_dr_k2 = equations(r + dr/2, m + k1_m/2, p + k1_p/2)
        k2_m = dr * dm_dr_k2
        k2_p = dr * dp_dr_k2
        
        # Step 3: Compute k3
        #print(f"Step 3:")
        dm_dr_k3, dp_dr_k3 = equations(r + dr/2, m + k2_m/2, p + k2_p/2)
        k3_m = dr * dm_dr_k3
        k3_p = dr * dp_dr_k3
        
        # Step 4: Compute k4
        #print(f"Step 4:")
        dm_dr_k4, dp_dr_k4 = equations(r + dr, m + k3_m, p + k3_p)
        k4_m = dr * dm_dr_k4
        k4_p = dr * dp_dr_k4

    # Update variables
        r += dr
        m += (k1_m + 2 * k2_m + 2 * k3_m + k4_m) / 6
        p += (k1_p + 2 * k2_p + 2 * k3_p + k4_p) / 6
        
        #print(f"r = {r}")
        #print("m = {:3e}".format(m))
        #print("p = {:3e}".format(p))
        
        # Update density using EOS
        rho = (p / K)**(1 / Gamma)
    
    # Store results
        r_list.append(r)
        m_list.append(m)
        p_list.append(p)
        rho_list.append(rho) 

    return r_list, m_list, p_list, rho_list

# Solve TOV equations
r_list, m_list, p_list, rho_list = rk4_integrate_TOV(p_c = (10**35)*(8.2624*10**(-45)), rho_c = (5*10**17)*(7.42592*10**(-28)), Gamma = 2, K = 1*10**5, r_max=350, dr=0.01)

print(f" ")
print(f"rho_list = {rho_list}")
print(f"r_list = {r_list}")
print(f"m_list = {m_list}")
print(f"p_list = {p_list}")
```

Defining the metric tensor for 3d spherical coordinates

```python
syms = sympy.symbols('r theta phi')
# define the metric for 3d spherical coordinates
metric = [[0 for i in range(3)] for i in range(3)]
metric[0][0] = 1
metric[1][1] = syms[0]**2
metric[2][2] = (syms[0]**2)*(sympy.sin(syms[1])**2)
# creating metric object
m_obj = MetricTensor(metric, syms)
print(f"syms = {syms}")
print(f"metric = {metric}")
print(f"m_obj = {m_obj}")
m_obj.tensor()
```
